```python
import pandas as pd
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
```

```python
#tokenize
t="""Hello I'm a girl studing in third year of college."""
tt=word_tokenize(t)
print(tt)
```

```python
#pos_tag
pt=nltk.pos_tag(tt)
print(pt)
```

```python
#stp wprds
stp=set(stopwords.words('english'))
fil=[]

for w in tt:
    if w not in stp:
        fil.append(w)
print(fil)
```

```python
#stemming
stem=PorterStemmer()
st=[stem.stem(word) for word in fil]
print(st)
```

```python
lemm=WordNetLemmatizer()
lm=[lemm.lemmatize(w) for w in fil]
print(lm)
```

```python
#tdf idf
from collections import Counter
tx="This is an example of TDF that is part of a7."

tok=tx.split()
tf=Counter(tok)

total=len(tok)
for w,count in tf.items():
    print("tdf of ",w," = ",count/total)
```

```python
#idf
import math
doc=["she is",
  "a girl",
   "she is also a woman"
  ]
n=len(doc)
w="is"
df=0
for d in doc:
    if w in d:
        df+=1
print("IDF: ", math.log(n/df))
```
